```python
import numpy as np
import matplotlib.pyplot as plt

shape = (530, 530, 3)

black_int16 = np.zeros(shape=shape, dtype=np.int16)
black_int32 = np.zeros(shape=shape, dtype=np.int32)

print("=== INT16 ===")
print("dtype :", black_int16.dtype)
print("shape :", black_int16.shape)
print("min   :", black_int16.min())
print("max   :", black_int16.max())
print("nbytes:", black_int16.nbytes)

print("\n=== INT32 ===")
print("dtype :", black_int32.dtype)
print("shape :", black_int32.shape)
print("min   :", black_int32.min())
print("max   :", black_int32.max())
print("nbytes:", black_int32.nbytes)

print("\n=== Perbandingan ===")
print("Ukuran int16 :", black_int16.nbytes, "bytes")
print("Ukuran int32 :", black_int32.nbytes, "bytes")
print("Perbandingan memori int32/int16:",
      black_int32.nbytes / black_int16.nbytes)

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(black_int16)
plt.title("int16")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(black_int32)
plt.title("int32")
plt.axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from skimage import io

path = '/content/drive/MyDrive/PCVK/ktm.jpeg'

# Baca menggunakan skimage
img_skimage = io.imread(path)

# Baca menggunakan OpenCV
img_bgr = cv2.imread(path)

# Konversi BGR menjadi RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Tentukan koordinat piksel yang sama
y, x = 100, 100

print("=== INFORMASI CITRA ===")
print("Shape skimage :", img_skimage.shape)
print("Shape cvtColor:", img_rgb.shape)

print("\n=== NILAI PIXEL ===")
print("Pixel skimage [R,G,B] :", img_skimage[y, x])
print("Pixel cv2 BGR [B,G,R]  :", img_bgr[y, x])
print("Pixel cv2 RGB [R,G,B]  :", img_rgb[y, x])

print("\n=== PERBANDINGAN ===")
print("skimage == cv2 RGB:",
      np.array_equal(img_skimage, img_rgb))

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_skimage)
plt.title("skimage.io.imread()")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.title("cv2.cvtColor(BGR → RGB)")
plt.axis('off')

plt.show()
```
